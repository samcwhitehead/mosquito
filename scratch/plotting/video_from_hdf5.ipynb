# Notebook to make a video from images in an hdf5 file

## Imports

```python
import os
import h5py
import cv2 

import numpy as np

from matplotlib import pyplot as plt
```

## Load data

```python
# where to look for data
data_path = r'C:\Users\samcw\Dropbox\flybratron\data\02_20251015\bag_and_hdf5\for_video'

for fn in os.listdir(data_path):
    if fn.endswith('.hdf5'):
        print(fn)
```

```python
# pick a data file and load images
data_filename = '251015_YAedes_YAedes_008_YPMV.hdf5'

with h5py.File(os.path.join(data_path, data_filename), "r") as f:
    imgs = f['images'][:]
    
```

## Make video

```python
# define video writer
video_filename = os.path.splitext(data_filename)[0] + '_video.avi'
nframes, height,width =imgs.shape
fps = 30

fourcc = cv2.VideoWriter_fourcc(*'XVID')
vw =cv2.VideoWriter(os.path.join(data_path, video_filename),fourcc,fps,(width,height))

# fourcc = cv2.VideoWriter_fourcc(*'mp4v')
# vw = cv2.VideoWriter('output.mp4', fourcc, fps, (width,height))
```

```python
# loop through images and write to file
for ith in range(nframes):
    frame = imgs[ith]
    frame_rgb = cv2.cvtColor(frame,cv2.COLOR_GRAY2RGB)
    vw.write(frame_rgb)

    cv2.imshow('frame', frame_rgb)
    if cv2.waitKey(1) == ord('q'):
        break
vw.release()
cv2.destroyAllWindows()
```
